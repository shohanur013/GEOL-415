```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('LA_AQS_2023.csv')
```

```python
print(df.head())
```

```python
# Get the list of column names
col_names = df.columns.tolist()

print("Column Names:", col_names)
```

```python
# Get unique names and sort them alphabetically
sorted_names = sorted(df['Parameter Name'].dropna().unique())

print("--- Alphabetical List of Parameters ---")
for name in sorted_names:
    print(f"- {name}")
```

```python
df_Ozone = df[(df['Parameter Name']=='Ozone')& (df['Duration Description']=='1 HOUR')]
df_Ozone.head()
```

```python
df_Ozone.describe()
```

```python
df_Ozone["Date (Local)"]
```

```python
df_Ozone["Date (Local)"] = pd.to_datetime(df_Ozone["Date (Local)"])
```

```python
df_Ozone["Date (Local)"]
```

```python
date = df_Ozone["Date (Local)"]
value = df_Ozone["Arithmetic Mean"]
```

```python
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(date, value)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(x=date, y=value)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(x=date, y=value)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(x=date, y=value)

plt.xlabel('Date', fontsize=12)
plt.ylabel('Ozone Concentration (ppm)', fontsize=12)

plt.show()
```

```python
df_Ozone["Arithmetic Mean (ppb)"] = df_Ozone["Arithmetic Mean"] * 1000
```

```python
highest_value = df_Ozone["Arithmetic Mean (ppb)"].max()
print("Highest Value:", highest_value)
```

```python
# 1. Find the row index of the highest value
highest_row_index = df_Ozone["Arithmetic Mean (ppb)"].idxmax()

# 2. Extract the date from that specific row
highest_date = df_Ozone.loc[highest_row_index, "Date (Local)"]
highest_value = df_Ozone.loc[highest_row_index, "Arithmetic Mean (ppb)"]

print(f"The highest value of {highest_value} occurred on: {highest_date}")
```

```python
above_35ppb = (df_Ozone["Arithmetic Mean (ppb)"] > 35).sum()
print(f"Total number of days with Ozone above 35 ppb: {above_35ppb}")
```
